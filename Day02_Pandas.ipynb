```python
import numpy as np
import pandas as pd
```

```python
myindex = ['USA', 'Canada', 'Mexico'] # create a list of countries
mydata = [1776, 1867, 1821] # create a list of data
myser = pd.Series(data = mydata, index = myindex) # join then and make it a series
myser
```

```python
ages = {'sam':5, 'frank':10, 'spike':7} # create a dict of index and data
ageser = pd.Series(ages) # this converts a dict into a pandas series
ageser
```

```python
q1 = {'japan':80, 'china':450, 'india':200, 'usa':250}
q2 = {'brazil':100, 'china':500, 'india':210, 'usa':260}
q1sales = pd.Series(q1)
q2sales = pd.Series(q2)
q1sales.keys() # returns all the keys
q2sales.keys() # returns all the keys
q1sales + q2sales # this gives NaN to those keys which are not in both the dicts
q1sales.add(q2sales, fill_value = 0) # fill_value fills existing missing(NaN) values we used here to give value as zero to keys which are not in the dict
```

```python
np.random.seed(101)
mydata = np.random.randint(0,101,(4,3))
mydata
myindex = ['CA','NY','AZ','TX']
mycolumns = ['Jan', 'Feb', 'Mar']
df = pd.DataFrame(data = mydata, index = myindex, columns = mycolumns) # this is how a DataFrame is created it has 3 input parameters data, index and columns
df
```

```python
df = pd.read_csv('tips.csv')
df.columns
df.index
# df.info() # gives the info of the whole DataFrame
df.describe().transpose() # describes all the statistic data of the DataFrame and .transpose() flips the df
df.head() # returns top 5 rows
df['tip_percentage'] = np.round(100 * df['tip'] / df['total_bill']) # adds a new column in the DataFrame and np.round() round's off the number
df = df.drop('tip_percentage', axis = 1) # .drop() this removes/drops a row or column depends on axis value
df = df.set_index('CC Number') # sets a column to an index
df.reset_index() # resets the df to default but new created things will exist but this doesn't save until i use df = df.reset_index()
df.iloc[0:4] # gives all the information of index at location 0
df.drop(4478071379779230, axis = 0) # this removes the index = 2 of axis = 0
```
