```python
# Conditional filtering
df = pd.read_csv('tips.csv')

df['total_bill'] > 20 # this returns a boolean df
newdf = df[df['total_bill'] > 20] # this returns the new dataframe filtered by the condition
newdf = newdf[newdf['sex'] == 'Male']
newdf = newdf[newdf['size'] >= 3]
newdf = newdf[newdf['smoker'] == 'Yes']
# both does the same work but difference in writing it
newdf_combined = df[(df['total_bill'] > 20) & (df['sex'] == 'Male') & (df['size'] >= 3) & (df['smoker'] == 'Yes')] # combining conditions
newdf_combined = newdf_combined[newdf_combined['day'].isin(['Sun', 'Sat', 'Fri'])] # .isin() is used for many options
```

```python
def last_four(num):
  return int(str(num)[-4:]) # this is how a function is created in python
df['CC Number'].apply(last_four) # here's how we implement it using .apply()
pd.Series(np.vectorize(last_four)(df['CC Number'])) # np.vectorize(#function)(df[]) this reduces time complexity as compared to .apply()

def bill(num):
  if(num > df['total_bill'].mean()): return '$$'
  else: return '$'
df['total_bill'].apply(bill)
df['yelp'] = pd.Series(np.vectorize(bill)(df['total_bill']))

# converting def function to lambda expression
def double(num):
  return num*2
# in lambda expression
lambda num: num*2

# example
df['total_bill'].apply(double)
df['total_bill'].apply(lambda num: num*2)

# applying custom function on more than 1 column
def quality(total_bill, tip):
  if tip/total_bill > 0.15:
    return "generous"
  else:
    return "other"

# np.vecctorize()() has a easy syntax and is faster
df['quality'] = df[['total_bill','tip']].apply(lambda df: quality(df['total_bill'], df['tip']),axis = 1)
df['quality'] = np.vectorize(quality)(df['total_bill'], df['tip']) # we will be using this way more
```
