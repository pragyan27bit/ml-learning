```python
np.arange(1,9,2) # creates an array from 0 till 9 excluded with a gap of 2
np.arange(0, 101, 20) # creates an array from 0 to 100 with a gap of 20
```

```python
np.zeros(5)
np.zeros((5,5)) # creates a matrix of 5 x 5 of all zeros
```

```python
np.ones(5) # creates a array of all elements as 1
```

```python
np.linspace(0,10,5) # creates an array starting from 0 till 10 included and divides into 5 parts
np.linspace(0, 10 , 11)
```

```python
np.eye(5) # creates a diagonal matrix
```

```python
np.random.rand(2,5) # creates an array of random numbers from 0 to 1
```

```python
np.random.randn(10) # creates an array of 10 random numbers
```

```python
np.random.randint(0, 100, 10) # creates an array of 10 random numbers between 0 and 100 excluded
```

```python
np.random.seed(42) # set a particular set of numbers
np.random.rand(4) # output is always same because of seed
```

```python
arr = np.random.randint(0, 100, 10)
arr.max() # returns the max value in the array
arr.argmax() # returns the index of the max value in the array
arr.min() # returns the min value in the array
arr.argmin() # returns the index of the min value in the array
```

```python
arr[5] # return the value at index 8
# arr[1:5] # returns a subarray from index 1 till 5 excluded
# arr[:5] # returns a subarray from index 0 till 5 excluded
# arr[5:] # returns a subarray from index 5 till end
# arrnew = arr.reshape(2,5)
# arrnew[0,3] # returns the value at given index
# boolarr = arr > 5 # return a boolean array for given condition
```

```python
arr = np.arange(0,10)
arr + 5
np.sqrt(arr) # returns the square root of every. value in the array
np.sin(arr)
np.log(arr)
arr.sum()
arr.mean()
arr.var() # returns variance of teh array
arr.std() # returns standard deviation of array
```

```python
arr2d = np.arange(0,25).reshape(5,5)
arr2d.sum(axis = 0) # axis = 0 is up to down or horizontal axis or across the row
arr2d.sum(axis = 1) # axis = 1 is left to right or vertical axis or across the column
```
